```python
x = df[['horsepower']]
y = df['price']
```

```python
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size = 0.2,random_state = 42)
```

```python
model = LinearRegression()
```

```python
print(f"Intercept of a model: {model.intercept_}")
print(f"Coefficient of a model : {model.coef_[0]}")

```

```python
# y = mx + c
# price = (156.16 × horsepower) - -3126.8699101927414
```

```python
y_pred = model.predict(x_test)
print(y_pred)
```

```python
mean_squared_error(y_test,y_pred)
```

```python
r2_score(y_test,y_pred)
```

```python
mean_absolute_error(y_test,y_pred)
```

```python
plt.scatter(x = y_test,y = y_pred)
plt.xlabel('y_test')
plt.ylabel('y_pred')
plt.show()
```
